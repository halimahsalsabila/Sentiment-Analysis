```python
import pandas as pd
import SentimentStrength

# Inisialisasi SentiStrength
senti = SentimentStrength()
senti.setLang("id")  # Gunakan Bahasa Indonesia

# Load data dari CSV
df = pd.read_csv("/content/crawling-data-day6-1.csv")

# Fungsi untuk menentukan label sentimen
def get_sentiment_label(text):
    try:
        score = senti.getSentiment(text)
        if score[0] > abs(score[1]):
            return 'positif'
        elif abs(score[1]) > score[0]:
            return 'negatif'
        else:
            return 'netral'
    except:
        return 'netral'  # fallback untuk teks kosong atau error

# Terapkan pelabelan ke kolom 'tweet'
df['label_sentimen'] = df['tweet'].apply(get_sentiment_label)

# Simpan hasilnya ke file baru
df.to_csv("tweets_berlabel.csv", index=False)

print("Labeling selesai. File disimpan sebagai 'tweets_berlabel.csv'")
```
